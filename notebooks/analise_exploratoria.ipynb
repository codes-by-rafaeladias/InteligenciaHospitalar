## DataSet
O conjunto de dados utilizado encontra-se disponível em: https://www.kaggle.com/datasets/sulphatet/diabetes-130us-hospitals-for-years-19992008. O repositório apresenta dois arquivos: registros de admissões de pacientes com Diabetes em 130 hospitais nos Estados Unidos durante os anos de 1999 a 2008 e a descrição dos tipos de admissão.

Quantidade de registros: 101766
Quantidade de variáveis: 50

## Extração dos Dados

O conjunto está originalmente no formato CSV e foi transformado em um DataFrame com o auxílio da biblioteca Python Pandas.
Inicialmente, foram realizadas análises sobre a quantidade de registros, número de variáveis, tipos das variáveis e quantidade de valores ausentes e únicos.

## Tratamento dos Dados

- Registros Duplicados e Ausentes: O conjunto não inclui registros duplicados, no entanto conta com poucos registros sobre os resultados de exames como Glicemia Sérica e HbA1c. Os registros ausentes foram preenchidos com o valor "Not registered".
- Valores Inválidos: Além disso, também verificou-se a presença de valores inválidos (números negativos) nas variáveis numéricas. Não foram encontrados.
- Mapeamento de Códigos e Descrições: Os códigos admission_type_id, discharge_disposition_id, admission_source_id foram mapeados para incluir suas descrições (registradas em outro arquivo utilitário CSV).
- Valores Ausentes nos IDs mapeados: Os valores NULL foram substituídos por "Not specified".
- Novas denominações para colunas: As colunas principais para a análise foram renomeadas.

## Exportação dos dados
Os dados foram exportados em um arquivo CSV e inseridos em uma tabela chamada "readmissoes_diabetes" em um banco de dados relacional no PostgreSQL.
